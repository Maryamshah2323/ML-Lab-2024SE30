## Task 9 · Grace marks
Using `marks`:
- **(a)** Add 5 grace marks but cap at 100 (`np.clip`).
- **(b)** Pass % (≥ 50) after grace.
- **(c)** Index of students who passed **only because of** the grace marks.
- **(d)** z-score of each student after grace.

```python
# T9a
marks_after_grace = np.clip(marks + 5, 0, 100)
print("T9a:", marks_after_grace)

# T9b
pass_percentage = np.mean(marks_after_grace >= 50) * 100
print("T9b:", pass_percentage)

# T9c
passed_only_after_grace = np.where((marks < 50) & (marks_after_grace >= 50))[0]
print("T9c:", passed_only_after_grace)

# T9d
z_scores = (marks_after_grace - np.mean(marks_after_grace)) / np.std(marks_after_grace)
print("T9d:", z_scores)
```

### Interpretation

The grace marks increased the students' marks by 5, without allowing the marks to exceed 100. The pass percentage shows the proportion of students who achieved at least 50 marks after the grace marks were added. Students who initially scored below 50 but passed after receiving grace marks are also identified. The Z-scores show how far each student's marks are from the average marks in terms of standard deviations.

## Task 10 · An image is an array
- **(a)** Start from `np.zeros((8, 8), dtype=int)` and draw the "7" from the lecture with slicing (white = 255, grey = 128).
- **(b)** Flip it left-right.
- **(c)** Invert it (`255 - img`).
- **(d)** Crop rows 1–4 and columns 3–6.
- **(e)** Brighten by 80, capped at 255.
- **(f)** Print its shape, dtype and number of white pixels.

*Optional:* `import matplotlib.pyplot as plt; plt.imshow(img, cmap='gray')`

```python
# T10a
img = np.zeros((8, 8), dtype=int)

# Draw number 7
img[0, :] = 255
img[1, 6] = 128
img[2, 5] = 128
img[3, 4] = 128
img[4, 3] = 128
img[5, 2] = 128
img[6, 1] = 128
img[7, 0] = 128

print("T10a:\n", img)

# T10b
flipped_img = np.fliplr(img)
print("T10b:\n", flipped_img)

# T10c
inverted_img = 255 - img
print("T10c:\n", inverted_img)

# T10d
cropped_img = img[1:5, 3:7]
print("T10d:\n", cropped_img)

# T10e
bright_img = np.clip(img + 80, 0, 255)
print("T10e:\n", bright_img)

# T10f
print("T10f Shape:", img.shape)
print("T10f Data type:", img.dtype)
print("T10f White pixels:", np.sum(img == 255))
```

### Interpretation

An 8×8 NumPy array was created to represent a grayscale image of the number 7. The image contains black, gray, and white pixels with intensity values of 0, 128, and 255, respectively. The image was flipped horizontally, inverted, cropped, and brightened using NumPy operations. The shape of the image is (8, 8), and its data type is integer. The image contains 8 white pixels.
